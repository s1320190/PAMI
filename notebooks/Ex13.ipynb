```python
import pandas as pd

def read_csv_with_encoding(file_name):
    encodings = ["utf-8", "utf-8-sig", "cp932", "shift_jis"]

    for encoding in encodings:
        try:
            df = pd.read_csv(file_name, encoding=encoding)
            print("使用した文字コード:", encoding)
            return df
        except UnicodeDecodeError:
            continue

    raise RuntimeError("CSVファイルを読み込めませんでした。")

original_df = read_csv_with_encoding(data_file)

print("データの行数と列数:", original_df.shape)
display(original_df.head())
```

```python
# NaNを0に置き換える
processed_df = original_df.copy()

processed_df = processed_df.fillna(0)

print("残っているNaN数:", processed_df.isna().sum().sum())

display(processed_df.head())
```

```python
# Timestamp列以外を取得
sensor_columns = [
    c for c in processed_df.columns
    if c.lower() != "timestamp"
]

# 数値型へ変換
for c in sensor_columns:
    processed_df[c] = pd.to_numeric(
        processed_df[c],
        errors="coerce"
    ).fillna(0)

# 100以上を0に置換
processed_df[sensor_columns] = processed_df[sensor_columns].mask(
    processed_df[sensor_columns] >= 100,
    0
)

print("最大値:", processed_df[sensor_columns].max().max())

display(processed_df.head())
```

```python
# Timestamp列を削除

if "Timestamp" in processed_df.columns:
    sensor_df = processed_df.drop(columns=["Timestamp"])
elif "timestamp" in processed_df.columns:
    sensor_df = processed_df.drop(columns=["timestamp"])
else:
    raise Exception("Timestamp column not found.")

print("データサイズ:", sensor_df.shape)

display(sensor_df.head())
```

```python
THRESHOLD = 15

transactions = []

for _, row in sensor_df.iterrows():

    transaction = []

    for column in sensor_df.columns:

        if row[column] >= THRESHOLD:
            transaction.append(column)

    transactions.append(transaction)

print("Transaction数:", len(transactions))

print("最初の5件")
for t in transactions[:5]:
    print(t)
```

```python
transaction_file = "PM24HeavyPollutionRecordingSensors.csv"

with open(transaction_file, "w", encoding="utf-8") as f:

    for t in transactions:

        if len(t) > 0:
            f.write(",".join(t) + "\n")

print("保存しました:", transaction_file)
```

```python
from PAMI.frequentPattern.basic import FPGrowth as alg

transaction_file = "PM24HeavyPollutionRecordingSensors.csv"

# 118件の80%以上、約95件以上に出現するパターン
min_sup = 0.8

fp = alg.FPGrowth(
    transaction_file,
    min_sup,
    ","
)

fp.mine()

patterns = fp.getPatternsAsDataFrame()

print("===== Result =====")
print("minSup:", min_sup)
print("Minimum support count:", int(118 * min_sup))
print("Number of patterns:", len(patterns))
print("Runtime:", fp.getRuntime(), "seconds")
print("Memory RSS:", fp.getMemoryRSS())

display(patterns.head(20))
```

```python
from collections import Counter

transaction_file = "PM24HeavyPollutionRecordingSensors.csv"

item_counts = Counter()
transaction_count = 0

with open(transaction_file, "r", encoding="utf-8") as f:
    for line in f:
        items = [
            item.strip()
            for item in line.strip().split(",")
            if item.strip()
        ]

        transaction_count += 1
        item_counts.update(set(items))

print("Transaction数:", transaction_count)
print("異なるセンサー数:", len(item_counts))

print("\n出現回数が多いセンサー上位20件")

for sensor, count in item_counts.most_common(20):
    rate = count / transaction_count

    print(
        sensor,
        "出現回数:", count,
        "出現割合:", f"{rate:.3f}"
    )
```

```python
max_sensor, max_count = item_counts.most_common(1)[0]
max_rate = max_count / transaction_count

print("最も多く出現したセンサー:", max_sensor)
print("最大出現回数:", max_count)
print("最大出現割合:", max_rate)
```

```python
min_sup = max_rate * 0.9

print("次に使用するminSup:", min_sup)
print(
    "最小支持度数:",
    int(transaction_count * min_sup)
)
```

```python
def pattern_length(pattern):
    if isinstance(pattern, (list, tuple, set)):
        return len(pattern)

    text = str(pattern).strip()

    # PAMIではタブ区切りで複数項目が入ることがある
    if "\t" in text:
        return len([x for x in text.split("\t") if x.strip()])

    # カンマ区切りの場合にも対応
    if "," in text:
        return len([x for x in text.split(",") if x.strip()])

    return 1 if text else 0


patterns["Length"] = patterns["Patterns"].apply(pattern_length)

print("最長パターンの長さ:", patterns["Length"].max())

display(
    patterns.sort_values(
        by=["Length", "Support"],
        ascending=[False, False]
    ).head(20)
)
```

```python
import pandas as pd

location_df = pd.read_csv(
    "stationInfo.csv",
    encoding="utf-8"
)

print(location_df.columns.tolist())
display(location_df.head())
```

```python
import re
import pandas as pd

# stationIdを8桁の文字列にそろえる
location_df["stationId"] = (
    location_df["stationId"]
    .astype(str)
    .str.replace(".0", "", regex=False)
    .str.zfill(8)
)

def parse_location(location):
    match = re.search(
        r"Point\(\s*([-\d.]+)\s+([-\d.]+)\s*\)",
        str(location),
        re.IGNORECASE
    )

    if match is None:
        return pd.Series([None, None])

    longitude = float(match.group(1))
    latitude = float(match.group(2))

    return pd.Series([longitude, latitude])

location_df[["Longitude", "Latitude"]] = location_df["Location"].apply(
    parse_location
)

display(location_df.head())
```

```python
longest_sensor_ids = [
    "13201020",
    "13201510",
    "13201040"
]

target_locations = location_df[
    location_df["stationId"].isin(longest_sensor_ids)
].copy()

print("抽出件数:", len(target_locations))
display(
    target_locations[
        ["stationId", "Address", "Latitude", "Longitude"]
    ]
)
```

```python
import plotly.express as px

fig = px.scatter_map(
    target_locations,
    lat="Latitude",
    lon="Longitude",
    hover_name="stationId",
    hover_data={
        "Address": True,
        "Latitude": True,
        "Longitude": True
    },
    zoom=10,
    height=600,
    title=(
        "Longest Frequent Pattern of Heavy PM2.5 Sensors "
        "(Length = 3, Support = 37)"
    )
)

fig.update_traces(
    marker={"size": 18}
)

fig.update_layout(
    map_style="open-street-map",
    margin={"r": 0, "t": 60, "l": 0, "b": 0}
)

fig.show()
```
